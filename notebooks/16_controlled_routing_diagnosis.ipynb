# Controlled routing diagnosis

## Research question
Can a router select MLP, graph, or attention *per sample* when the correct route is generated with the sample?

## Hypothesis
A supervised router should learn observable construction labels. Comparing natural descriptors against explicit metadata distinguishes an information problem from an optimization problem.

## Benchmark and oracle

The generator creates independent, locally correlated, and endpoint-correlated sequence families. Its family ID is the oracle route. Natural descriptors are computed from the actual sequence; the explicit one-hot family is a diagnostic control only. See `docs/superpowers/specs/2026-09-04-controlled-routing-diagnosis-design.md` for the precise contract.

```python
from pathlib import Path
from neuroforge.training import run_routing_diagnosis
from neuroforge.visualization import save_routing_diagnosis_figures

result = run_routing_diagnosis(Path('../results/metrics/notebook_controlled_routing'), seeds=(11, 23, 37), epochs=40)
save_routing_diagnosis_figures(result, '../figures/notebook_controlled_routing')
result['aggregate']
```

## Diagnostics

Compare `route_accuracy_mean`, per-seed confusion matrices, utilization, active modules, and `router_batch_ms`. Soft conditions evaluate all specialists and are therefore not physical compute savings. ## Interpretation, limitations, next experiment

Explicit metadata success establishes router learnability; imperfect natural routing identifies an observation/descriptor bottleneck. This notebook does not demonstrate that the Phase 1 neural blocks are individually specialized, nor that routing improves neural latency. The next experiment must train capacity-matched primitive baselines per family before reintegrating them with the learned router.
